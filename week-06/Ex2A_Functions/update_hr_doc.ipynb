## Lab 3. Employee data 
This lab exercise makes updates to information extracted from 
an employee database. 

```python
hr_list = [ 
('0123', 'HR', 'Rebecca Yang', '69000'),  
('0121', 'IT', 'Mark Blick', '67000'),  
('0068', 'IT', 'Kahna Larsen', '112000'),  
('0234', 'OPS', 'Jim Smith', '54000') 
] 
```

```python
# Tuples are immutable
# First, find Mark's index in the list
mark_index = next(i for i, emp in enumerate(hr_list) if 'Mark' in emp[2])

# Unpack the old tuple, replace the name, rebuild the tuple
emp_id, dept, name, salary = hr_list[mark_index]
hr_list[mark_index] = (emp_id, dept, 'Mark Blick-Hawley', salary)
```

```python
print(hr_list)
```

```python
# Find Jim's index in the list
jim_index = next(i for i, emp in enumerate(hr_list) if 'Jim' in emp[2])

# Unpack the old tuple, replace dept and salary, rebuild the tuple
emp_id, dept, name, salary = hr_list[jim_index]
hr_list[jim_index] = (emp_id, 'CS', name, '60000')
```

```python
print(hr_list)
```

```python
# Loop through each employee 
for emp in hr_list:
    emp_id, dept, name, salary = emp

    # salary to int so we can apply comma 
    print(f"{emp_id} | {dept} | {name} | {int(salary):,}")
```
