## Lab 1. Working with names – parsing strings 

```python
test_name = 'John Doe'
```

```python
def parse_and_display (name):
        for i in range (len(name)):
            if name[i] == ' ': 
                first = name[:i] 
                last = name[i+1:] 
                print(f"Full name: {name}") 
                print(f"First name: {first}") 
                print(f"Last name:  {last}")  
parse_and_display (test_name)
```

```python
parse_and_display (test_list[0])
```

```python
parse_and_display (test_list[2])
```

```python
def parse_and_display2(name):
    parts = name.strip().split()

    if len(parts) == 1:
        first = parts[0]
        middle = None
        last = None
    elif len(parts) == 2:
        first = parts[0]
        middle = None
        last = parts[1]
    elif len(parts) == 3:
        first = parts[0]
        middle = parts[1]
        last = parts[2]
    else:
        print("Error: please enter 1, 2, or 3 words only.")
        return

    print(f"Full name:   {name.strip()}")
    print(f"First name:  {first}")
    if middle:
        print(f"Middle name: {middle}")
    print(f"Last name:   {last if last else 'N/A'}")
```

```python
test_list_2= ["Lorde", "Billie Eilish ", "Megan Thee Stallion "]
```
