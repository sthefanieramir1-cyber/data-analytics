# Exercise 2.A Using built-in functions 
## Lab 2. Parsing strings – part codes

```python
# Examples: 
# ACME:123-L supplied by Acme, part #123, size large 
# DI:12345-M supplied by DI, part #12345, size medium 
# ACE:1-12 
# supplied by Ace, part #1, size 12 
```

```python
def get_supplier_code(part_code):
    return part_code[:part_code.index(':')]
```

```python
def get_product_num(part_code):
    colon_pos = part_code.index(':')
    dash_pos = part_code.index('-')
    return part_code[colon_pos + 1 : dash_pos]
```

```python
def get_size(part_code):
    return part_code[part_code.index('-') + 1:]
```

```python
part1 = "ACME:123-L"
part2 = "DI:12345-M"
part3 = "ACE:1-12"
```

```python
print(f"For {part1}, the supplier is {get_supplier_code(part1)}. "
      f"Part number {get_product_num(part1)} is available in size {get_size(part1)}.")
```

```python
print(f"For {part2}, the supplier is {get_supplier_code(part2)}. "
      f"Part number {get_product_num(part2)} is available in size {get_size(part2)}.")
```

```python
print(f"For {part3}, the supplier is {get_supplier_code(part3)}. "
      f"Part number {get_product_num(part3)} is available in size {get_size(part3)}.")
```

```python
# Store all part codes in a list
part_codes = ["ACME:123-L", "DI:12345-M", "ACE:1-12"]

# Loop through each part code and call all three functions
for code in part_codes:
    supplier = get_supplier_code(code)
    product  = get_product_num(code)
    size     = get_size(code)
    print(f"For {code}, the supplier is {supplier}. "
          f"Part number {product} is available in size {size}.")
```
