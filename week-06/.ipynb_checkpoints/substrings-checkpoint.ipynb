## Lab 1. Working with names – parsing strings

```python
test_name = 'Peter Pan'
```

```python
def parse_and_display(name): 
    for i in range(len(name)): 
        if name[i] == ' ': 
            first = name[:i] 
            last = name[i+1:] 
            print(f"Full name: {name}") 
            print(f"First name: {first}") 
            print(f"Last name:  {last}")
```

```python
parse_and_display(test_name) 
```

# Classes

## Defining a class

```python
class SimpleDog: 
    '''Simple model for a dog''' 

    def __init__(self, name, age, he_she): 
        self.name = name 
        self.age = age 
        self.he_she = he_she 
    
    def sit(self):
        print(f'{self.name} is now sitting')
```

```python
Sthef_dog  = SimpleDog("Benji", 4, "He")
```

```python
print(f"The dog's name is {Sthef_dog.name}")
print(f"{Sthef_dog.he_she} is {Sthef_dog.age}")
```

```python
class Lumber:
    '''Pieces of Lumbers'''
    def __init__(self,width, wood ="pine") :
        self.wood = wood
        self.width = width
        self.lenght = 96
```

```python
class SimpleDog: 
    '''Simple model for a dog''' 

    def __init__(self, name, age, he_she): 
        self.name = name 
        self.age = age 
        self.he_she = he_she 
        self.behavior = 'good'
    
    def sit(self):
        print(f'{self.name} is now sitting')
    def describe_behavior(self):
        return f'{self.he_she} is a {self.behavior} dog'

    def birthday(self) : 
        if str(self.age).lower() == 'less than 1' :
            self.age = 1
        else :
            self.age += 1
        print(f'{self.name} is now {self.age}')
        return f'Happy brithday {self.name}' 
```

```python
Sthef_dog  = SimpleDog("Benji", 4, "He")
Sara_dog  = SimpleDog("Peanuts", "less than 1", "She")
Alex_dog  = SimpleDog("Rico Suave", 4, "He")
```

```python
print(Sthef_dog.birthday())
print(Sara_dog.birthday())
print(Alex_dog.birthday())
```
