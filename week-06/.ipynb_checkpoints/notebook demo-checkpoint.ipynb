```python
walamrt_shopping = ["conditioner #P123 on store", "Shampoo #P0234 out of stuck", "Carrots #P2255 out of stock"] 

shopping = []

for i in walamrt_shopping:
    if "out of stock" not in i: 
        shopping.append(i[i.find("P"):i.find("P")+20])
print(shopping)
```

```python
walmart_shopping = ["conditioner #P123 on store", "Shampoo #P0234 out of stock", "Carrots #P2255 out of stock"]
shopping = []
for i in walmart_shopping:
    if "out of stock" not in i:
        shopping.append(i[i.find("P"):].split()[0])
print(shopping)
```

# String Formatting

```python
txt1 = "I ate {burgers} burgers and now I feel {feeling}"
print(txt1.format(burgers = 5, feeling = "like a balloon"))
```

```python
txt = "I ate {slices:.2f} slices of pizza"
print(txt.format(slices = 3))
```

```python
txt = "My dog weighs {weight:.1f} lbs"  # 1 decimal
print(txt.format(weight = 24.678))

txt2 = "My dog weighs {weight:.3f} lbs"  # 3 decimals
print(txt2.format(weight = 24.678))
```

```python
final_receipt = "Total plus tax is ${total:.2f}"  # template with $ sign

food_bill = {"entree1": 23, "entree2": 18, "drink1": 4.00, "drink2": 7.00}

subtotal = sum(food_bill.values())  # adds all values → 52

tax = subtotal * .08  # 8% tax → 4.16

total = subtotal + tax  # 52 + 4.16 → 56.16

print(final_receipt.format(total = total))
```

```python
print(math.fabs(6.7)) # absolute value : Number is its distance from zero, 
print(math.fabs(-6.7))

# Distance & Difference When you want to know how far apart two numbers are, the order shouldn't matter.
# Subtracting in the wrong order gives a negative result, 
# so absolute value ensures you always get a clean, positive distance.
```
