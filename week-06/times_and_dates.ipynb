## Lab 2 - Exercise 2.B: Working with Datetime Functions


```python
import datetime
```

```python
# Today
```

```python
today = datetime.datetime.now()
```

```python
# %A = full weekday, %B = full month name, %d = day, %Y = 4-digit year
print(f"Today is {today.strftime('%A, %B %d, %Y')}")
```

```python
# My Birthday
```

```python
birthday = datetime.date(2002, 7, 18)
```

```python
# %m = month number, %d = day, %Y = year
print(f"My birthday is {birthday.strftime('%m/%d/%Y')}")
```

```python
# Next 90 Days
```

```python
# timedelta adds 90 days to today's date
ninety_d = today + datetime.timedelta(days=90)
```

```python
print(f"90 days from today is {ninety_d.strftime('%B %d, %Y')}")
```

```python
# Time for Dinner
```

```python
# datetime.time(hour, minute) — using 24hr values, strftime handles AM/PM
dinner_time = datetime.time(19, 30)
```

```python
# %I = 12-hour clock, %M = minutes, %p = AM/PM
print(f"Let's meet for dinner at {dinner_time.strftime('%I:%M %p')}")
```
