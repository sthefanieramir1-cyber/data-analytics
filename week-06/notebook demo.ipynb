# String Formatting

```python
txt1 = "I ate {burgers} burgers and now I feel {feeling}"
print(txt1.format(burgers = 5, feeling = "like a balloon"))
```

```python
txt = "I ate {slices:.2f} slices of pizza"
print(txt.format(slices = 3))
```

```python
txt = "My dog weighs {weight:.1f} lbs"  # 1 decimal
print(txt.format(weight = 24.678))

txt2 = "My dog weighs {weight:.3f} lbs"  # 3 decimals
print(txt2.format(weight = 24.678))
```

```python
final_receipt = "Total plus tax is ${total:.2f}"  # template with $ sign

food_bill = {"entree1": 23, "entree2": 18, "drink1": 4.00, "drink2": 7.00}

subtotal = sum(food_bill.values())  # adds all values → 52

tax = subtotal * .08  # 8% tax → 4.16

total = subtotal + tax  # 52 + 4.16 → 56.16

print(final_receipt.format(total = total))
```

```python
import random

# Single random number between 1 and 100
a = random.randint(1, 100)
print(a)

# Random decimal number between 0 and 1
b = random.random()
print(b)

# Random decimal in a custom range
c = random.uniform(1.0, 10.0)
print(c)

# Pick a random item from a list
fruits = ["apple", "banana", "cherry", "mango"]
d = random.choice(fruits)
print(d)

# Pick multiple random items from a list (no repeats)
e = random.sample(fruits, 2)
print(e)

# Shuffle a list in place
random.shuffle(fruits)
print(fruits)

# Random number from a specific set of values
f = random.randrange(0, 100, 10)  # picks from 0,10,20...90
print(f)
```

```python
# using it inside a name funtion 
def discount_calc2(pct_off): 
    return lambda price: price * (1 - pct_off) 
```

# Error Handling 
## Working with Try / Except 

```python
first_num = float(input("% to calculate as decimal :" ))

pct_as_dec = first_num / 100 

print(pct_as_dec)
```

```python
try : 
        first_num = float(input("% to calculate as decimal :" ))
        pct_as_dec = first_num / 100 
except : 
        print("Invalid input. Try again")
else :
        print(pct_as_dec)
```

```python
try : 
        first_num = float(input("% to calculate as decimal :" ))
        pct_as_dec = first_num / 100 
except Value_error: 
        print("Invalid input: Must be a number. Try again")
else :
        print(pct_as_dec)
```
