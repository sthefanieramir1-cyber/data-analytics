```python
import numpy as np
import pandas as pd
```

## Series  objects

```python
aa = [1, 2, 3, 4]
ab = (5, 6, 7, 8)
ac = {9, 10, 11, 12}
ad = {'first': 13, 'second': 14, 'third': 15, 'fourth': 16}
```

## Series  create and print

```python
ser1 = pd.Series(aa)
ser2 = pd.Series(ab)
ser3 = pd.Series(ac)
ser4 = pd.Series(ad)

print("ser1 (from list):");  print(ser1)
print("ser2 (from tuple):"); print(ser2)
print("ser3 (from set):");   print(ser3)
print("ser4 (from dict):");  print(ser4)
```

```python
# Create df1 from ser1
df1 = pd.DataFrame(ser1)
df1

# Add two columns via insert()
df1.insert(1, 'col2', ser2)
df1.insert(2, 'col3', ser3)
df1

# Add col4 from ser4 (dict-based — mismatched index causes NaN)
df1['col4'] = ser4
df1
# ser4 has string index ('first','second',...) vs integer index in df1,
# so pandas can't align values → col4 fills entirely with NaN.

# Add a new row at index 4 using loc
df1.loc[4] = [5, 9, 12, 0]
df1
```
