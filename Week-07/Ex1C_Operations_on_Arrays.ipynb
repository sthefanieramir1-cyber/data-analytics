## Sample Arrays

```python
# Reusing r_1 through r_5 from Exercise 1.B
r_1 = np.random.randint(1, 10, size=5)
r_2 = np.random.randint(1, 10, size=(2, 5))
r_3 = np.random.uniform(1, 10, size=(3, 4))
r_4 = np.random.randint(1, 10, size=(2, 5))
r_5 = np.random.randint(1, 10, size=(3, 3))
print(r_1, r_2, r_3, r_4, r_5, sep='\n')
```

## Permutations

```python
print('r_4 before shuffle:\n', r_4)
```

```python
np.random.shuffle(r_4)  # modifies r_4 in place
print('r_4 after shuffle:\n', r_4)
```

The elements were shuffled along axis 0, meaning the rows swapped positions. The values inside each row stayed in their original order  only the rows themselves moved.

```python
new_shuffle = np.random.permutation(r_4)  # returns new array, r_4 unchanged
print('new_shuffle:\n', new_shuffle)
print('r_4 (unchanged):\n', r_4)
```

**shuffle vs permutation:** Use `shuffle` to reorder the original array in place. 
Use `permutation` to get a shuffled copy without changing the original.

## Mathematical Operations

```python
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_2:\n', r_2, '| shape:', r_2.shape)
result = np.add(r_1, r_2)
print('r_1 + r_2:\n', result, '| shape:', result.shape)
```

```python
print('r_1:', r_1, '| shape:', r_1.shape)
print('r_3:\n', r_3, '| shape:', r_3.shape)

try:
    result = np.add(r_1, r_3)
    print('r_1 + r_3:\n', result, '| shape:', result.shape)
except ValueError as e:
    print('Error:', e)
# r_1 is shape (5,) and r_3 is shape (3,4) columns don't match so broadcasting fails
```

```python
print('r_2:\n', r_2, '| shape:', r_2.shape)
print('r_4:\n', r_4, '| shape:', r_4.shape)
result = np.add(r_2, r_4)
print('r_2 + r_4:\n', result, '| shape:', result.shape)
```

```python
# r_1 (5,) + r_4 (2,5)  works, r_1 broadcasts across both rows of r_4
try:
    print('r_1 + r_4:\n', np.add(r_1, r_4))
except ValueError as e:
    print('r_1 + r_4 error:', e)

# r_2 (2,5) + r_4 (2,5)  same shape, works fine
try:
    print('r_2 + r_4:\n', np.add(r_2, r_4))
except ValueError as e:
    print('r_2 + r_4 error:', e)

# r_3 (3,4) + r_4 (2,5)  incompatible shapes, fails
try:
    print('r_3 + r_4:\n', np.add(r_3, r_4))
except ValueError as e:
    print('r_3 + r_4 error:', e)
```

```python
print(np.add(np.array([1]), r_4))
```

```python
print(np.add(np.array([[1]]), r_4))
```

```python
print(np.add(np.array([1,1,1,1,1]), r_4))
```

```python
# [[1],[1]] shape (2,1)  broadcasts across the 5 columns
try:
    print(np.add(np.array([[1],[1]]), r_4))
except Exception as e:
    print('Error:', e)
```

**How broadcasting works:** NumPy stretches a smaller array to fit a larger one when 
dimensions are compatible (sizes match or one is 1). The result always takes the shape 
of the larger array.

## Rounding

```python
print('r_3:\n', r_3)
```

```python
print(np.trunc(r_3))    # drops the decimal, rounds toward zero
```

```python
print(np.fix(r_3))      # same as trunc, rounds toward zero
```

```python
print(np.floor(r_3))    # rounds down toward negative infinity
```

```python
print(np.ceil(r_3))     # rounds up toward positive infinity
```

```python
print(np.round(r_3, 2)) # standard rounding to 2 decimal places
```

## Grocery Problem

```python
# [vegetables, fruits, snacks, beverages]
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print('bag1:', bag1)
print('bag2:', bag2)
```

```python
totals = np.sum([bag1, bag2], axis=1)
print(f'The first bag contains {totals[0]} items, and the second bag contains {totals[1]} items.')
```

```python
by_category = np.add(bag1, bag2)
print(f'There are {by_category[0]} vegetables, {by_category[1]} fruits, {by_category[2]} boxes of snacks, and {by_category[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = np.array([5, 4, 6, 4, 2])  # Mon–Fri
```

```python
ben = np.array([0, 5, 5, 3, 4])  # 0 = missed Monday
```

```python
carla = np.array([3, 6, 0, 3, 4])  # 0 = missed Wednesday
```

```python
all_knocks = np.array([ali, ben, carla])  # rows = people, columns = days
print('all_knocks:\n', all_knocks)
```

```python
per_person = np.sum(all_knocks, axis=1)   # sum each row = weekly total per person
group_total = np.sum(all_knocks)
print(f'Ali knocked on {per_person[0]} doors, Ben on {per_person[1]}, Carla on {per_person[2]}.')
print(f'The group collectively knocked on {group_total} doors.')
```

```python
daily_totals = np.sum(all_knocks, axis=0)  # total doors per day across the whole group
rolling_daily = np.cumsum(daily_totals)    # rolling total of doors day by day
rolling_votes = rolling_daily * 0.5        # 1 estimated vote per every 2 door knocks

days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday']
for day, votes in zip(days, rolling_votes):
    print(f'{day}: {int(votes)} estimated votes turned out so far')
```
