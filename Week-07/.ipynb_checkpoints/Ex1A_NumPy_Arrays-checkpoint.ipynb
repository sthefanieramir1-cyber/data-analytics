# Exercise 1.A NumPy: Manipulating Arrays

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

**First loop** (`for x in arrA`): `x` is a whole row  prints 2 lines, each a doubled array like `[22, 24, 26]`.

**Second loop** (`for y in x`): `y` is a single element  prints 6 individual doubled numbers, one per line.

The difference: the first loop works **row by row**, the second drills into **each individual value**.

```python
# x is the whole row, so x*2 doubles the full row  not just the current y value
# that's why I get the same doubled row printed 3 times per row (once per element)
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
```

## Joining and Stacking Arrays

```python
# stacking arrA on top of itself  axis=0 means adding more rows
joined_array_1 = np.concatenate((arrA, arrA), axis=0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
try:
    stacked_array_2 = np.stack((joined_array_1, joined_array_2))
except ValueError as e:
    print(e)
# this fails because np.stack needs both arrays to be the same shape
# joined_array_1 is (4,3) and joined_array_2 is (2,6)  they don't match
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
try:
    another_2 = np.concatenate((joined_array_1, arrA), axis=1)
except ValueError as e:
    print(e)
# this fails because axis=1 needs matching row counts
# joined_array_1 has 4 rows, arrA only has 2  can't join them side by side
```

```python
hstacked = np.hstack((arrA, arrA))
print(hstacked)  # same as axis=1  adds more columns

vstacked = np.vstack((arrA, arrA))
print(vstacked)  # same as axis=0  adds more rows

dstacked = np.dstack((arrA, arrA))
print(dstacked)  # goes 3D  pairs up elements along a new depth axis
```

## Splitting Arrays

```python
# splitting joined_array_1 into 2 equal parts
splitsky = np.split(joined_array_1, 2)
print(splitsky)
# I get 2 arrays, each 2 dimensional (shape 2x3)
```

```python
# joined_array_1 has 4 rows so I need array_split to handle the uneven division
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
# I get 3 arrays, each 2 dimensional (two with 2 rows, one with 1 row)
```

## Searching and Replacing Array Values

```python
# using where() to check each value if even keep it, if odd replace with 'odd'
# % 2 == 0 is my condition to test for even numbers
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
# creating arrB and sorting it  np.sort sorts each row smallest to largest
arrB = np.array([[7, 6], [13, 5], [7, 3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
# filtering arrB to only get values less than 7
# z is a boolean mask True where condition is met
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
# same idea but flipped  now I want values greater than or equal to 7
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
