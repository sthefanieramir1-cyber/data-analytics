```python
import numpy as np
```

## Generating Random Numbers

```python
# random integers between 1 and 20, 2 elements
r_1 = np.random.randint(1, 20, size=2)
print(r_1)
```

```python
# random integers between 1 and 20, 1 row 2 columns
r_2 = np.random.randint(1, 20, size=(1, 2))
print(r_2)
```

```python
# random floats between 0 and 1, 2 rows 1 column
r_3 = np.random.random(size=(2, 1))
print(r_3)
```

```python
# random integers between 20 and 100, 4 rows 5 columns
r_4 = np.random.randint(20, 100, size=(4, 5))
print(r_4)
```

```python
# random floats between 0 and 1, 4 rows 5 columns
r_5 = np.random.random(size=(4, 5))
print(r_5)
```

## Generating Random Choices from an Array

```python
# storing 12 classmate names in a 1D array
students = np.array(['Alice', 'Bob', 'Carlos', 'Diana', 'Eva', 'Frank',
                     'Grace', 'Henry', 'Isla', 'James', 'Karen', 'Leo'])
```

```python
# picking one random student to call on
call_on = np.random.choice(students)
print(call_on)
```

```python
# randomly assigning 12 students into 3 rows, 4 columns (3 groups of 4)
breakout_groups = np.random.choice(students, size=(3, 4))
print(breakout_groups)
```

## Generating a Normal Distribution

```python
# generating a bell curve with mean=10, std=5, size 4 rows 5 columns
bell_curve = np.random.normal(loc=10, scale=5, size=(4, 5))
print(bell_curve)
```
